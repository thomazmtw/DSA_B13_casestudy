```python
# part F

import sqlite3
import pandas as pd
import xml.etree.ElementTree as ET
```

```python
# 1. Load and Clean Datasets
df_pop = pd.read_csv('/content/drive/MyDrive/dsa /new data sets/population.csv')
df_gdp = pd.read_excel('/content/drive/MyDrive/dsa /new data sets/gdp.xlsx').rename(columns={'Country': 'country'})
df_net = pd.read_json('/content/drive/MyDrive/dsa /new data sets/internet_users.json')
tree = ET.parse('/content/drive/MyDrive/dsa /new data sets/literacy_rate.xml')
root = tree.getroot()
data = [{'country': c.get('name'), 'literacy_rate': float(c.find('literacy_rate').text)} for c in root.findall('country')]
df_lit = pd.DataFrame(data)
```

```python
# 2. Merge Datasets & Engineer Calculated Columns
df_merged = df_pop.merge(df_gdp, on='country') \
                  .merge(df_net, on='country') \
                  .merge(df_lit, on='country')

df_merged['internet_penetration'] = ((df_merged['internet_users'] / df_merged['population']) * 100).round(2)
df_merged['gdp_per_capita'] = ((df_merged['GDP'] * 1000000) / df_merged['population']).round(2)
```

```python
# 3. Create SQLite Database and Table
conn = sqlite3.connect('country_statistics.db')
df_merged.to_sql('country_statistics', conn, if_exists='replace', index=False)

def run_sql(query):
    return pd.read_sql_query(query, conn)
```

```python
# Q11: Top 5 countries by GDP
print("--- Question 11: Top 5 countries by GDP ---")
q11 = """
SELECT country, GDP
FROM country_statistics
ORDER BY GDP DESC
LIMIT 5;
"""
print(run_sql(q11), "\n")
```

```python
# Q12: Average literacy rate
print("--- Question 12: Average Literacy Rate ---")
q12 = """
SELECT ROUND(AVG(literacy_rate), 2) AS average_literacy_rate
FROM country_statistics;
"""
print(run_sql(q12), "\n")
```

```python
# Q13: Countries having literacy rate above 90%
print("--- Question 13: Countries with Literacy Rate > 90% ---")
q13 = """
SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90
ORDER BY literacy_rate DESC;
"""
print(run_sql(q13), "\n")
```

```python
# Q14: Top 5 countries with highest internet penetration
print("--- Question 14: Top 5 Countries with Highest Internet Penetration ---")
q14 = """
SELECT country, internet_penetration
FROM country_statistics
ORDER BY internet_penetration DESC
LIMIT 5;
"""
print(run_sql(q14), "\n")
```

```python
# Q15: Calculate total population
print("--- Question 15: Total Population ---")
q15 = """
SELECT SUM(population) AS total_population
FROM country_statistics;
"""
print(run_sql(q15), "\n")
```

```python
# Q16: Country with highest GDP per capita
print("--- Question 16: Country with Highest GDP Per Capita ---")
q16 = """
SELECT country, gdp_per_capita
FROM country_statistics
ORDER BY gdp_per_capita DESC
LIMIT 1;
"""
print(run_sql(q16), "\n")
```

```python
# Q17: Export final merged dataset as CSV file
df_merged.to_csv('final_country_statistics.csv', index=False)
print("--- Question 17: Dataset exported as 'final_country_statistics.csv' --- \n")
```

```python
# Q18: Sort dataset by GDP per Capita (Descending) and display Top 10
print("--- Question 18: Top 10 Countries by GDP Per Capita ---")
q18 = """
SELECT country, gdp_per_capita
FROM country_statistics
ORDER BY gdp_per_capita DESC
LIMIT 10;
"""
print(run_sql(q18))
```
