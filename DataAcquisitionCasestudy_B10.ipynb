```python
#task 4

import xml.etree.ElementTree as ET

# Read XML file
tree = ET.parse('/content/drive/MyDrive/dsa /new data sets/literacy_rate.xml')
root = tree.getroot()

data = []
for country_elem in root.findall('country'):
    name = country_elem.get('name')
    lit_rate = float(country_elem.find('literacy_rate').text)
    data.append({'country': name, 'literacy_rate': lit_rate})

df_lit = pd.DataFrame(data)
print(df_lit.head())
```

```python
# Load datasets
df_pop = pd.read_csv('/content/drive/MyDrive/dsa /new data sets/population.csv')
df_gdp = pd.read_excel('/content/drive/MyDrive/dsa /new data sets/gdp.xlsx')
df_net = pd.read_json('/content/drive/MyDrive/dsa /new data sets/internet_users.json')

# Parse XML dataset
tree = ET.parse('/content/drive/MyDrive/dsa /new data sets/literacy_rate.xml')
root = tree.getroot()
data = [{'country': c.get('name'), 'literacy_rate': float(c.find('literacy_rate').text)} for c in root.findall('country')]
df_lit = pd.DataFrame(data)

# Dictionary of datasets
datasets = {
    "Population Dataset (population.csv)": df_pop,
    "GDP Dataset (gdp.xlsx)": df_gdp,
    "Internet Users Dataset (internet_users.json)": df_net,
    "Literacy Rate Dataset (literacy_rate.xml)": df_lit
}

# Display exploration details
for name, df in datasets.items():
    print(f"=== {name} ===")
    print(f"Number of rows: {df.shape[0]}")
    print(f"Number of columns: {df.shape[1]}")
    print(f"Column names: {list(df.columns)}")
    print("Data types:")
    for col, dt in df.dtypes.items():
        print(f"  - {col}: {dt}")
    print("Missing values:")
    for col, val in df.isnull().sum().items():
        print(f"  - {col}: {val}")
    print(f"Duplicate records: {df.duplicated().sum()}\n")
```

```python
#part c task 6

# Rename 'Country' to 'country' in GDP dataset
df_gdp = df_gdp.rename(columns={'Country': 'country'})

df_gdp
```

```python
#task 7
# Convert numeric columns to standard numeric data types
df_pop['population'] = df_pop['population'].astype('int64')
df_gdp['GDP'] = df_gdp['GDP'].astype('float64')
df_net['internet_users'] = df_net['internet_users'].astype('int64')
df_lit['literacy_rate'] = df_lit['literacy_rate'].astype('float64')
```

```python
# task 8
# Inner join on common column 'country' correctly merging all four dataframes
df_merged = df_pop.merge(df_gdp, on='country') \
                  .merge(df_net, on='country') \
                  .merge(df_lit, on='country')
```

```python
# 1. Internet Penetration (%) = (internet_users / population) * 100
df_merged['internet_penetration'] = ((df_merged['internet_users'] / df_merged['population']) * 100).round(2)

# 2. GDP Per Capita = (GDP in Millions * 1,000,000) / population
df_merged['gdp_per_capita'] = ((df_merged['GDP'] * 1000000) / df_merged['population']).round(2)

df_merged.head()
```

```python
#task 10 part D

import sqlite3

conn = sqlite3.connect('country_statistics.db')

df_merged.to_sql('country_statistics', conn, if_exists='replace', index=False)
```

```python
#task 11
# Retrieve first 5 records from SQLite table
cursor = conn.cursor()
cursor.execute("SELECT * FROM country_statistics LIMIT 5")
rows = cursor.fetchall()
for row in rows:
    print(row)
```
